### Models Integration with OpenAI, Gemini and Groq

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
```

```python
from langchain.chat_models import init_chat_model
model = init_chat_model("google_genai:gemini-2.5-flash")
# another way to write the above line of code - 
# model = init_chat_model("gemini-2.5-flash", model_provider="google_genai")
model
```

```python
#invoke the model
response=model.invoke("Hello, how are you?")
response
```

```python
response.content
```

```python
response2 = model.invoke("Why do parrots talk?")
response2.content
```

```python
## Using chatOpenAI  (OPENAI API)
from langchain_openai import ChatOpenAI
model2 = ChatOpenAI(model="gpt-4.1")
response3 = model2.invoke("Hello, how are you?")
response3
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
model = ChatGoogleGenerativeAI(model="gemini-2.5-flash")
response4 = model.invoke("Why do parrots talk?")
response4
```
